```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
# from langchain_google_vertexai import VertexAIEmbeddings
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=3000, chunk_overlap=250)
# embeddings = GoogleGenerativeAIEmbeddings(model="models/text-embedding-003")
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")
# embeddings = VertexAIEmbeddings(model="text-embedding-004")
```

```python
# embeddings = GoogleGenerativeAIEmbeddings(model="models/text-embedding-004",transport = 'grpc_asyncio')
# vector = embeddings.embed_query("hello, world!")
# vector[:5]
```

```python
loader = DirectoryLoader(
        "/kaggle/input/resume/",
        glob="**/*.pdf",
        loader_cls=PyPDFLoader
    )
documents = loader.load()
chunks = text_splitter.split_documents(documents)
vector_store = FAISS.from_documents(chunks, embeddings)
```

```python
from langchain.chains import RetrievalQAWithSourcesChain
```

```python
# Iterate through all documents in the vector store
for i in range(vector_store.index.ntotal):
    # Get the docstore ID corresponding to the FAISS index
    doc_id = vector_store.index_to_docstore_id[i]
    # Retrieve the document from the docstore
    doc = vector_store.docstore.search(doc_id)
    if doc:
        print(f"Index: {i}")
        print(f"Content: {doc.page_content}")
        print(f"Metadata: {doc.metadata}")
        print("-" * 50)
    else:
        print(f"No document found for index {i}")
```

```python
# Step 5: Set up the RetrievalQAWithSourcesChain
qa_chain = RetrievalQAWithSourcesChain.from_chain_type(
    llm=llm,
    chain_type="stuff",  # Simple concatenation of retrieved docs
    retriever=vector_store.as_retriever(),  # Retrieve top 2 similar documents
    return_source_documents=True  # Return the source documents in the response
)
```

```python
query = """Extract the response in provided format.:

List[
    {"employer_name":string,
    "employment start year":string,
    "employment end year":string,
    "designation":string,
    "place_of_employment":string}
]    
"""
response = qa_chain({"question": query})

# Step 7: Print the response
print("Answer:", response["answer"])
print("Sources:", response["sources"])
# print("Source Documents:", [doc.metadata["source"] for doc in response["source_documents"]])
```
